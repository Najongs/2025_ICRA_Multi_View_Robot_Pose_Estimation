```python
import os
import glob
import json

search_dir = "../DREAM_real/panda-3cam_azure"
output_dir = "DREAM_to_DREAM/panda-3cam_azure"

# search_dir = "../DREAM_real/panda-3cam_kinect360"
# output_dir = "DREAM_to_DREAM/panda-3cam_kinect360"

# search_dir = "../DREAM_real/panda-3cam_realsense"
# output_dir = "DREAM_to_DREAM/panda-3cam_realsense"

# search_dir = "../DREAM_real/panda-orb"
# output_dir = "DREAM_to_DREAM/panda-orb"

os.makedirs(output_dir, exist_ok=True)
json_files = glob.glob(os.path.join(search_dir, '**', '*.json'), recursive=True)

print(f"총 {len(json_files)}개의 JSON 파일을 찾았습니다.")

for json_path in json_files:
    with open(json_path, "r", encoding="utf-8") as f:
        sample = json.load(f)

    if 'objects' not in sample:
        intrinsics = sample['camera_settings'][0]['intrinsic_settings']
        fx = intrinsics['fx']
        fy = intrinsics['fy']
        cx = intrinsics['cx']
        cy = intrinsics['cy']

        k_matrix = [
            [fx, 0.0, cx],
            [0.0, fy, cy],
            [0.0, 0.0, 1.0]
        ]

        meta_to_add = {
            "K": k_matrix,
        }

        print(f"✅ Successfully checked the 'meta' data from '{json_path}'")

    else:
        path_without_ext = os.path.splitext(json_path)[0]
        image_path = f"{path_without_ext}.rgb.jpg"
        
        meta_to_add['image_path'] = os.path.join('../dataset', image_path.split('../')[1])
        sample['meta'] = meta_to_add

        original_filename = os.path.basename(json_path)
        output_json_path = os.path.join(output_dir, original_filename)

        with open(output_json_path, 'w', encoding='utf-8') as f:
            json.dump(sample, f, indent=4)
```
